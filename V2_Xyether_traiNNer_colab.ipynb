```python
# @title #**⚡  Dataset Pre-Processing (For Faster training)**

import os
import sys
import re
import random
import numpy as np
import cv2
import torch
import shutil
import types
from torch.nn import functional as F
from torch.utils.data import DataLoader
from tqdm.auto import tqdm

hr_dataset_roots = "/content/datasets/HFA2k/hfa2k" # @param {type:"string"}
output_dir = "/content/datasets/Preprocessed" # @param {type:"string"}

Noise_Level = "Soft" # @param ["Soft", "Medium", "Strong"]
Compression_Level = "Soft" # @param ["Soft", "Medium", "Strong"]
Blur_Level = "Soft" # @param ["Soft", "Medium", "Strong"]
APISR_USM_Preprocessing = True # @param {type:"boolean"}
Zip_and_Upload_to_Drive = True # @param {type:"boolean"}
Force_Rechop = False # @param {type:"boolean"}
Force_Redegrade = False # @param {type:"boolean"}

trainner_dir = "/content/drive/MyDrive/training ai ups/traiNNer-redux"
gt_size = 192
Crop_Stride = 192
Cleanup_Temp_Crops = False

sys.path.append(trainner_dir)

class DotDict(dict):
    def __getattr__(self, key):
        if key not in self:
            if re.search(r'resize_mode_list\d*', key): return ['bilinear', 'bicubic']
            if re.search(r'resize_mode_prob\d*', key): return [0.5, 0.5]
            if re.search(r'kernel_list\d*', key): return ['iso', 'aniso']
            if re.search(r'kernel_prob\d*', key): return [0.5, 0.5]
            if re.search(r'blur_sigma\d*', key): return [0.2, 3.0]
            if re.search(r'betag_range\d*', key): return [0.5, 4.0]
            if re.search(r'betap_range\d*', key): return [1.0, 2.0]
            if re.search(r'_prob\d*$', key): return 1.0
            if re.search(r'_range\d*$', key):
                if 'jpeg' in key: return [30, 95]
                return [0.15, 1.5]
            if 'debug' in key: return False
            if 'type' in key: return 'keep'
            return None
        return self.get(key)
    __setattr__ = dict.__setitem__
    __delattr__ = dict.__delitem__

def make_dotdict(d):
    if isinstance(d, dict):
        return DotDict({k: make_dotdict(v) for k, v in d.items()})
    elif isinstance(d, list):
        return [make_dotdict(v) for v in d]
    return d

try:
    from traiNNer.data.realesrgan_dataset import RealESRGANDataset
    from traiNNer.utils.config import Config
    from traiNNer.models.realesrgan_model import RealESRGANModel
    from traiNNer.models.sr_model import SRModel

    Config.get_manual_seed = classmethod(lambda cls: 1024)
    Config.get_config = classmethod(lambda cls: (make_dotdict({'manual_seed': 1024}), None))
except ImportError:
    print("❌ Failed to import TraiNNer modules. Make sure your Drive is mounted and the path exists.")
    sys.exit(1)

temp_hr_dir = "/content/datasets/Temp_HR_Crops"
if Force_Rechop and os.path.exists(temp_hr_dir):
    print("⚠️ Force Rechop enabled. Clearing existing Temp crops...")
    shutil.rmtree(temp_hr_dir)
os.makedirs(temp_hr_dir, exist_ok=True)

existing_temp_crops = [f for f in os.listdir(temp_hr_dir) if f.endswith('.png')]

if len(existing_temp_crops) > 0:
    print(f"⏩ Smart Resume: Found {len(existing_temp_crops)} existing crops in Temp folder! Skipping chopping phase.")
    total_crops = len(existing_temp_crops)
else:
    dataset_paths = [p.strip() for p in hr_dataset_roots.split(",") if p.strip()]
    valid_exts = ('.png', '.jpg', '.jpeg', '.webp')
    all_images = []
    for path in dataset_paths:
        if os.path.exists(path):
            for root, _, files in os.walk(path):
                for f in files:
                    if f.lower().endswith(valid_exts):
                        all_images.append(os.path.join(root, f))

    print(f"🔪 Chopping {len(all_images)} images into {gt_size}x{gt_size} crops...")
    total_crops = 0
    for img_path in tqdm(all_images, desc="Chopping Images"):
        img = cv2.imread(img_path)
        if img is None: continue
        h, w = img.shape[:2]

        for y in range(0, h - gt_size + 1, Crop_Stride):
            for x in range(0, w - gt_size + 1, Crop_Stride):
                crop = img[y:y+gt_size, x:x+gt_size]
                cv2.imwrite(os.path.join(temp_hr_dir, f"{total_crops:06d}.png"), crop)
                total_crops += 1

    print(f"✅ Generated {total_crops} perfect {gt_size}x{gt_size} crops based on your dataset!")

if total_crops == 0:
    print("❌ Error: 0 crops generated. Please check your dataset paths.")
    sys.exit(1)

def get_scale(level):
    if level == "Strong": return 1.0
    if level == "Medium": return 0.7
    return 0.4 # Soft

n_scale, c_scale, b_scale = get_scale(Noise_Level), get_scale(Compression_Level), get_scale(Blur_Level)

strong = {
    "resize_prob": [0.2, 0.7, 0.1], "resize_range": [0.15, 1.5],
    "gaussian_noise_prob": 0.5, "noise_range": [1, 30],
    "poisson_scale_range": [0.05, 3], "gray_noise_prob": 0.4,
    "jpeg_range": [30, 95], "second_blur_prob": 0.8,
    "resize_prob2": [0.3, 0.4, 0.3], "resize_range2": [0.3, 1.2],
    "gaussian_noise_prob2": 0.5, "noise_range2": [1, 25],
    "poisson_scale_range2": [0.05, 2.5], "gray_noise_prob2": 0.4,
    "jpeg_range2": [30, 95]
}

def scale_prob(p, scale):
    if isinstance(p, list):
        scaled_up, scaled_down = p[0] * scale, p[1] * scale
        return [scaled_up, scaled_down, 1.0 - (scaled_up + scaled_down)]
    return p * scale

def scale_range(r, scale, center=None, reverse=False):
    if reverse: return [int(95 - (95 - r[0]) * scale), r[1]]
    elif center is not None: return [center - (center - r[0]) * scale, center + (r[1] - center) * scale]
    else:
        low, high = r[0], r[0] + (r[1] - r[0]) * scale
        if isinstance(r[0], int) and isinstance(r[1], int): return [int(low), int(high)]
        return [low, high]

opt = {
    'name': 'TrainDataset', 'type': 'realesrgandataset',
    'dataroot_gt': [temp_hr_dir],
    'gt_size': gt_size, 'scale': 2, 'use_hflip': True, 'use_rot': True,
    'io_backend': {'type': 'disk'}, 'use_usm': False,

    'queue_size': 0, 'thicklines_prob': 0,

    'datasets': {'train': {'gt_size': gt_size + 64}},

    'blur_prob': 1.0, 'noise_prob': 1.0, 'jpeg_prob': 1.0,
    'noise_prob2': 1.0, 'jpeg_prob2': 1.0,
    'color_jitter_prob': 0.0, 'color_jitter_pt_prob': 0.0,

    'kernel_range': [7, 21],
    'kernel_list': ['iso', 'aniso', 'generalized_iso', 'generalized_aniso', 'plateau_iso', 'plateau_aniso'],
    'kernel_prob': [0.45, 0.25, 0.12, 0.03, 0.12, 0.03],
    'sinc_prob': 0.1 * b_scale, 'blur_sigma': [0.2, 3.0 * b_scale], 'betag_range': [0.5, 4], 'betap_range': [1, 2],

    'resize_prob': scale_prob(strong['resize_prob'], c_scale),
    'resize_range': scale_range(strong['resize_range'], c_scale, center=1.0),

    'gaussian_noise_prob': scale_prob(strong['gaussian_noise_prob'], n_scale),
    'noise_range': scale_range(strong['noise_range'], n_scale),
    'poisson_scale_range': scale_range(strong['poisson_scale_range'], n_scale),
    'gray_noise_prob': scale_prob(strong['gray_noise_prob'], n_scale),
    'jpeg_range': scale_range(strong['jpeg_range'], c_scale, reverse=True),

    'second_blur_prob': scale_prob(strong['second_blur_prob'], b_scale),
    'kernel_range2': [7, 21],
    'kernel_list2': ['iso', 'aniso', 'generalized_iso', 'generalized_aniso', 'plateau_iso', 'plateau_aniso'],
    'kernel_prob2': [0.45, 0.25, 0.12, 0.03, 0.12, 0.03],
    'sinc_prob2': 0.1 * b_scale, 'blur_sigma2': [0.2, 1.5 * b_scale], 'betag_range2': [0.5, 4], 'betap_range2': [1, 2],

    'resize_prob2': scale_prob(strong['resize_prob2'], c_scale),
    'resize_range2': scale_range(strong['resize_range2'], c_scale, center=1.0),

    'gaussian_noise_prob2': scale_prob(strong['gaussian_noise_prob2'], n_scale),
    'noise_range2': scale_range(strong['noise_range2'], n_scale),
    'poisson_scale_range2': scale_range(strong['poisson_scale_range2'], n_scale),
    'gray_noise_prob2': scale_prob(strong['gray_noise_prob2'], n_scale),
    'jpeg_range2': scale_range(strong['jpeg_range2'], c_scale, reverse=True),
    'final_sinc_prob': 0.8 * b_scale,
    'final_kernel_range': [7, 21]
}

opt = make_dotdict(opt)

print("⚙️ Initializing TraiNNer's Native Degradation Pipeline...")
original_init = SRModel.__init__

def mocked_init(self, opt):
    self.opt = opt
    self.device = torch.device('cuda')
    self.is_train = True
    self.lq = None
    self.gt = None

    self.memory_format = torch.preserve_format
    self.amp = False
    self.autocast_dtype = torch.float32
    self.accum_iters = 1
    self.batch_augment = None # Bypasses Mixture of Augments (MoA) checks

SRModel.__init__ = mocked_init
pipeline = RealESRGANModel(opt)
SRModel.__init__ = original_init # Restore immediately

pipeline._dequeue_and_enqueue = types.MethodType(lambda self: None, pipeline)

class GPU_APISR_USM(torch.nn.Module):
    def __init__(self, device):
        super().__init__()
        self.device = device
        k1d = cv2.getGaussianKernel(7, 0)
        self.register_buffer('usm_kernel', torch.FloatTensor(np.dot(k1d, k1d.T)).unsqueeze(0).unsqueeze(0))
        sm_k = cv2.getGaussianKernel(3, 0.5)
        self.register_buffer('smooth_kernel', torch.FloatTensor(np.dot(sm_k, sm_k.T)).unsqueeze(0).unsqueeze(0))

    def filter2D_pt(self, img, kernel):
        c, pad = img.size(1), kernel.size(-1) // 2
        return F.conv2d(F.pad(img, (pad, pad, pad, pad), mode='reflect'), kernel.repeat(c, 1, 1, 1), groups=c)

    def apply_apisr(self, img, weight, thresh_val, passes):
        thresh = thresh_val / 255.0
        def usm_pass(x):
            blur = self.filter2D_pt(x, self.usm_kernel)
            res = x - blur
            soft_mask = self.filter2D_pt((torch.abs(res) > thresh).float(), self.usm_kernel)
            return soft_mask * torch.clamp(x + weight * res, 0, 1) + (1.0 - soft_mask) * x
        sharp = usm_pass(img)
        for _ in range(passes): sharp = usm_pass(sharp)
        return torch.clamp(sharp, 0, 1)

    def forward(self, gt_ctx):
        gt_ctx = gt_ctx.to(self.device, non_blocking=True)
        gt_ctx = self.filter2D_pt(gt_ctx, self.smooth_kernel)
        gt_ctx = self.apply_apisr(gt_ctx, 0.3, 10, 2)
        return gt_ctx

print("⚙️ Initializing TraiNNer's Native Dataloader...")
dataset = RealESRGANDataset(opt)
dataloader = DataLoader(dataset, batch_size=16, shuffle=False, num_workers=0, pin_memory=True)
usm_processor = GPU_APISR_USM(torch.device('cuda')).cuda() if APISR_USM_Preprocessing else None

hr_out = os.path.join(output_dir, "HR")
lr_out = os.path.join(output_dir, "LR")

if Force_Redegrade:
    print("⚠️ Force Redegrade enabled. Clearing previous output pairs...")
    if os.path.exists(hr_out): shutil.rmtree(hr_out)
    if os.path.exists(lr_out): shutil.rmtree(lr_out)

os.makedirs(hr_out, exist_ok=True)
os.makedirs(lr_out, exist_ok=True)

saved_count = len([f for f in os.listdir(hr_out) if f.endswith('.png')])

if saved_count > 0:
    print(f"⏩ Smart Resume: Found {saved_count} already degraded pairs in output folder. Fast-forwarding dataloader...")

print(f"🚀 Degrading patches using PyTorch...")
pbar = tqdm(total=total_crops, initial=saved_count, desc="Degrading & Saving")
processed_in_loop = 0

for batch in dataloader:
    batch_size_current = batch['gt'].shape[0]

    if processed_in_loop + batch_size_current <= saved_count:
        processed_in_loop += batch_size_current
        continue

    start_idx = 0
    if processed_in_loop < saved_count:
        start_idx = saved_count - processed_in_loop

    pad_val = 32
    batch['gt'] = F.pad(batch['gt'], (pad_val, pad_val, pad_val, pad_val), mode='reflect')

    with torch.no_grad():
        pipeline.feed_data(batch)
        gts = pipeline.gt[start_idx:]
        lqs = pipeline.lq[start_idx:]

        pad_lq = pad_val // 2
        gts = gts[..., pad_val:-pad_val, pad_val:-pad_val]
        lqs = lqs[..., pad_lq:-pad_lq, pad_lq:-pad_lq]

        if usm_processor is not None:
            gts = usm_processor(gts)

    gts, lqs = gts.cpu().numpy(), lqs.cpu().numpy()

    for i in range(gts.shape[0]):
        gt_img = np.transpose(gts[i], (1, 2, 0)) * 255.0
        lq_img = np.transpose(lqs[i], (1, 2, 0)) * 255.0

        cv2.imwrite(os.path.join(hr_out, f"{saved_count:06d}.png"), cv2.cvtColor(gt_img.astype(np.uint8), cv2.COLOR_RGB2BGR))
        cv2.imwrite(os.path.join(lr_out, f"{saved_count:06d}.png"), cv2.cvtColor(lq_img.astype(np.uint8), cv2.COLOR_RGB2BGR))

        saved_count += 1
        pbar.update(1)

    processed_in_loop += batch_size_current

    if saved_count >= total_crops:
        break

if Cleanup_Temp_Crops:
    print(f"🧹 Cleaning up temporary chopped folder...")
    shutil.rmtree(temp_hr_dir)

print(f"✅ Preprocessing fully completed! {saved_count} perfectly matched image pairs saved to {output_dir}")

if Zip_and_Upload_to_Drive:
    print(f"📦 Zipping dataset...")
    zip_path = "/content/Preprocessed_Dataset.zip"
    shutil.make_archive(zip_path.replace('.zip', ''), 'zip', output_dir)
    print(f"✅ Zipped successfully to {zip_path}")

    drive_dir = "/content/drive/MyDrive"
    if os.path.exists(drive_dir):
        print(f"🚀 Uploading to Google Drive...")
        dest = os.path.join(drive_dir, "Preprocessed_Dataset.zip")
        shutil.move(zip_path, dest)
        print(f"✅ Uploaded successfully to {dest}")
    else:
        print(f"⚠️ Google Drive is not connected. The zip has been kept at {zip_path}")
```

```python
# @title #**⚙️Model Training Setup**

project_name = "2x_DigitalArtV3-HFA2k-ApisrOTF-Compact-Xyether9K" # @param {type:"string"}
# @markdown ### 📁 **Dataset Roots**
# @markdown Enter a single path or multiple paths separated by commas.
hr_dataset_roots = "/content/datasets/DigitalArtV3/digital_art_v3, /content/datasets/HFA2k/hfa2k" # @param {type:"string"}
Model_Architecture = "compact" # @param ["compact", "ultracompact", "superultracompact"]

# @markdown ### 🎛️ **Degradation Settings**
Noise_Level = "Soft" # @param ["Soft", "Medium", "Strong"]
Compression_Level = "Soft" # @param ["Soft", "Medium", "Strong"]
Blur_Level = "Soft" # @param ["Soft", "Medium", "Strong"]
APISR_USM_Preprocessing = True # @param {type:"boolean"}

# @markdown ### ⏱️ **Training Length**
Training_Iterations = 25000 # @param {type:"integer"}

import os
import random
import shutil
import cv2
import numpy as np
import matplotlib.pyplot as plt
import sys
import glob
import subprocess
from IPython.display import display

trainner_dir = "/content/drive/MyDrive/training ai ups/traiNNer-redux"

dataset_paths = [p.strip() for p in hr_dataset_roots.split(",") if p.strip()]
if not dataset_paths:
    dataset_paths = [hr_dataset_roots]

yaml_dataset_list = str(dataset_paths)

pretrains = {
    "compact": ("1tpdHD3T1yPd6986H0uscOX-pMd9PkJER", "compact_pretrain.pth"),
    "ultracompact": ("18ZD_UxFqUuGWG8nHeoLgZ0GUnDMTZgat", "ultracompact_pretrain.pth"),
    "superultracompact": ("1POCpRAIACKHvIjYTbWGjaiXGmDYWN85t", "superultracompact_pretrain.pth")
}

pretrain_id, pretrain_filename = pretrains[Model_Architecture]
pretrain_path = f"{trainner_dir}/experiments/pretrains/{pretrain_filename}"

if not os.path.exists(pretrain_path):
    print(f"⏬ Downloading {Model_Architecture} pretrain model from Google Drive...")
    os.makedirs(os.path.dirname(pretrain_path), exist_ok=True)
    subprocess.run(["gdown", "--id", pretrain_id, "-O", pretrain_path])

network_g_yaml = f"""network_g:
  type: {Model_Architecture}"""

base_val_dir = os.path.dirname(dataset_paths[0]) if dataset_paths else "/content/datasets"
val_hr_dir = os.path.join(base_val_dir, "Auto_VAL_HR")
val_lr_dir = os.path.join(base_val_dir, "Auto_VAL_LR")
os.makedirs(val_hr_dir, exist_ok=True)
os.makedirs(val_lr_dir, exist_ok=True)

valid_exts = ('.png', '.jpg', '.jpeg', '.webp')
all_images = []

for path in dataset_paths:
    if os.path.exists(path):
        for root, _, files in os.walk(path):
            for f in files:
                if f.lower().endswith(valid_exts):
                    all_images.append(os.path.join(root, f))

if len(os.listdir(val_hr_dir)) < 3 and len(all_images) >= 3:
    val_images = random.sample(all_images, 3)
    for img_path in val_images:
        shutil.copy(img_path, os.path.join(val_hr_dir, os.path.basename(img_path)))

val_files = [f for f in os.listdir(val_hr_dir) if f.lower().endswith(valid_exts)][:3]

def get_scale(level):
    if level == "Strong": return 1.0
    if level == "Medium": return 0.7
    if level == "Soft": return 0.4
    return 1.0

noise_scale = get_scale(Noise_Level)
comp_scale = get_scale(Compression_Level)
blur_scale = get_scale(Blur_Level)

def scale_prob(p, scale):
    if isinstance(p, list):
        scaled_up = p[0] * scale
        scaled_down = p[1] * scale
        keep = 1.0 - (scaled_up + scaled_down)
        return f"[{scaled_up:.2f}, {scaled_down:.2f}, {keep:.2f}]"
    return f"{p * scale:.2f}"

def scale_range(r, scale, center=None, reverse=False):
    if reverse: # jpeg
        low = 95 - (95 - r[0]) * scale
        return f"[{int(low)}, {r[1]}]"
    elif center is not None:
        low = center - (center - r[0]) * scale
        high = center + (r[1] - center) * scale
        return f"[{low:.2f}, {high:.2f}]"
    else:
        low = r[0]
        high = r[0] + (r[1] - r[0]) * scale
        if isinstance(r[0], int) and isinstance(r[1], int):
            return f"[{int(low)}, {int(high)}]"
        return f"[{low:.2f}, {high:.2f}]"

strong = {
    "resize_prob": [0.2, 0.7, 0.1],
    "resize_range": [0.15, 1.5],
    "gaussian_noise_prob": 0.5,
    "noise_range": [1, 30],
    "poisson_scale_range": [0.05, 3],
    "gray_noise_prob": 0.4,
    "jpeg_range": [30, 95],
    "second_blur_prob": 0.8,
    "resize_prob2": [0.3, 0.4, 0.3],
    "resize_range2": [0.3, 1.2],
    "gaussian_noise_prob2": 0.5,
    "noise_range2": [1, 25],
    "poisson_scale_range2": [0.05, 2.5],
    "gray_noise_prob2": 0.4,
    "jpeg_range2": [30, 95]
}

yaml_otf = f"""
    blur_kernel_size: 21
    kernel_list: ['iso', 'aniso', 'generalized_iso', 'generalized_aniso', 'plateau_iso', 'plateau_aniso']
    kernel_prob: [0.45, 0.25, 0.12, 0.03, 0.12, 0.03]
    sinc_prob: {0.1 * blur_scale:.2f}
    blur_sigma: [0.2, {3.0 * blur_scale:.2f}]
    betag_range: [0.5, 4]
    betap_range: [1, 2]

    resize_prob: {scale_prob(strong['resize_prob'], comp_scale)}
    resize_range: {scale_range(strong['resize_range'], comp_scale, center=1.0)}

    gaussian_noise_prob: {scale_prob(strong['gaussian_noise_prob'], noise_scale)}
    noise_range: {scale_range(strong['noise_range'], noise_scale)}
    poisson_scale_range: {scale_range(strong['poisson_scale_range'], noise_scale)}
    gray_noise_prob: {scale_prob(strong['gray_noise_prob'], noise_scale)}

    jpeg_range: {scale_range(strong['jpeg_range'], comp_scale, reverse=True)}

    second_blur_prob: {scale_prob(strong['second_blur_prob'], blur_scale)}
    blur_kernel_size2: 21
    kernel_list2: ['iso', 'aniso', 'generalized_iso', 'generalized_aniso', 'plateau_iso', 'plateau_aniso']
    kernel_prob2: [0.45, 0.25, 0.12, 0.03, 0.12, 0.03]
    sinc_prob2: {0.1 * blur_scale:.2f}
    blur_sigma2: [0.2, {1.5 * blur_scale:.2f}]
    betag_range2: [0.5, 4]
    betap_range2: [1, 2]

    resize_prob2: {scale_prob(strong['resize_prob2'], comp_scale)}
    resize_range2: {scale_range(strong['resize_range2'], comp_scale, center=1.0)}

    gaussian_noise_prob2: {scale_prob(strong['gaussian_noise_prob2'], noise_scale)}
    noise_range2: {scale_range(strong['noise_range2'], noise_scale)}
    poisson_scale_range2: {scale_range(strong['poisson_scale_range2'], noise_scale)}
    gray_noise_prob2: {scale_prob(strong['gray_noise_prob2'], noise_scale)}

    jpeg_range2: {scale_range(strong['jpeg_range2'], comp_scale, reverse=True)}

    final_sinc_prob: {0.8 * blur_scale:.2f}
"""
usm_bool = "True" if APISR_USM_Preprocessing else "False"

if val_files:
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    fig.suptitle(f"Preview: HR vs Approx. Degraded LR", fontsize=16)

    for i, img_name in enumerate(val_files):
        hr_path = os.path.join(val_hr_dir, img_name)
        img_hr = cv2.imread(hr_path)
        if img_hr is None: img_hr = np.random.randint(0, 255, (256, 256, 3), dtype=np.uint8)

        if APISR_USM_Preprocessing:
            blur = cv2.GaussianBlur(img_hr, (21, 21), 0)
            img_hr = cv2.addWeighted(img_hr, 1.45, blur, -0.45, 0)

        h, w = img_hr.shape[:2]
        img_lr = cv2.resize(img_hr, (w // 2, h // 2), interpolation=cv2.INTER_CUBIC)
        lh, lw = img_lr.shape[:2]

        if random.random() < strong['second_blur_prob'] * blur_scale:
            sigma = random.uniform(0.2, 1.5 * blur_scale)
            img_lr = cv2.GaussianBlur(img_lr, (5, 5), sigma)

        resample_prob = strong['resize_prob'][0] * comp_scale + strong['resize_prob'][1] * comp_scale
        if random.random() < resample_prob:
            sx, sy = random.uniform(0.4, 1.6), random.uniform(0.4, 1.6)
            rw, rh = max(8, int(lw * sx)), max(8, int(lh * sy))
            img_lr = cv2.resize(img_lr, (rw, rh), interpolation=cv2.INTER_CUBIC)
            img_lr = cv2.resize(img_lr, (lw, lh), interpolation=cv2.INTER_CUBIC)

        jpeg_lb = 95 - (95 - strong['jpeg_range'][0]) * comp_scale
        if random.random() < strong['jpeg_range'][1] / 100.0 * comp_scale:
            b_factor = random.uniform(jpeg_lb / 100.0, 0.95)
            bh, bw = max(8, int(lw * b_factor)), max(8, int(lh * b_factor))
            bh, bw = bh - bh % 8, bw - bw % 8
            tmp = cv2.resize(img_lr, (bw, bh), interpolation=cv2.INTER_NEAREST)
            img_lr = cv2.resize(tmp, (lw, lh), interpolation=cv2.INTER_LINEAR)

        if random.random() < strong['gaussian_noise_prob'] * noise_scale:
            n_lvl = random.uniform(1, 30 * noise_scale)
            noise = np.random.normal(0, n_lvl, img_lr.shape).astype(np.float32)
            img_lr = np.clip(img_lr.astype(np.float32) + noise, 0, 255).astype(np.uint8)

        cv2.imwrite(os.path.join(val_lr_dir, img_name), img_lr)

        axes[0, i].imshow(cv2.cvtColor(img_hr, cv2.COLOR_BGR2RGB))
        axes[0, i].set_title(f"Target HR {i+1}")
        axes[0, i].axis('off')

        axes[1, i].imshow(cv2.cvtColor(img_lr, cv2.COLOR_BGR2RGB))
        axes[1, i].set_title(f"Degraded LR {i+1}")
        axes[1, i].axis('off')

    plt.tight_layout()
    display(fig)
    plt.close()

phase1_total = int(Training_Iterations * 0.8)
phase2_total = Training_Iterations - phase1_total

def get_latest_model(exp_dir):
    models = glob.glob(os.path.join(exp_dir, "models", "net_g_*.safetensors")) + glob.glob(os.path.join(exp_dir, "models", "net_g_*.pth"))
    if not models: return None
    def get_iter(p):
        try: return int(os.path.basename(p).split('_')[-1].split('.')[0])
        except: return -1
    models.sort(key=get_iter)
    return models[-1]

phase1_exp = os.path.join(trainner_dir, "experiments", f"{project_name}_Phase1")
phase2_exp = os.path.join(trainner_dir, "experiments", f"{project_name}_Phase2")

phase1_done = False
if get_latest_model(phase2_exp):
    phase1_done = True
elif get_latest_model(phase1_exp):
    latest_iter = int(os.path.basename(get_latest_model(phase1_exp)).split('_')[-1].split('.')[0])
    if latest_iter >= phase1_total: phase1_done = True

def generate_config(phase_num):
    is_p1 = phase_num == 1
    c_name = f"{project_name}_Phase1" if is_p1 else f"{project_name}_Phase2"
    c_total = phase1_total if is_p1 else phase2_total
    c_batch = 2 if is_p1 else 4

    m1, m2, m3, m4 = int(c_total*0.5), int(c_total*0.8), int(c_total*0.9), int(c_total*0.95)
    milestones_list = f"[{m1}, {m2}, {m3}, {m4}]"

    if is_p1:
        losses_yaml = """
    - type: l1loss
      loss_weight: 1.0"""
        network_d_yaml = ""
        optim_d_yaml = ""
        pretrain = pretrain_path
    else:
        losses_yaml = """
    - type: mssimloss
      loss_weight: 0.5
    - type: perceptualloss
      criterion: charbonnier
      loss_weight: 0.01
    - type: hsluvloss
      criterion: charbonnier
      loss_weight: 1.0
    - type: ganloss
      gan_type: vanilla
      loss_weight: 0.08"""
        network_d_yaml = """
network_d:
  type: dunet"""
        optim_d_yaml = """
  optim_d:
    type: AdamW
    lr: !!float 2e-4
    weight_decay: 0
    betas: [0.9, 0.99]"""
        pretrain = f"{trainner_dir}/experiments/{project_name}_Phase1/models/net_g_{phase1_total}.safetensors"

    return f"""
name: {c_name}
auto_resume: true
scale: 2
use_amp: true
amp_bf16: false
use_channels_last: true
fast_matmul: true
use_compile: false
compile_mode: default
num_gpu: 1

datasets:
  train:
    name: Train Dataset
    type: realesrgandataset
    dataroot_gt: {yaml_dataset_list}
    gt_size: 128
    use_hflip: true
    use_rot: true
    num_worker_per_gpu: 2
    batch_size_per_gpu: {c_batch}
    accum_iter: 1
    prefetch_mode: cuda
    pin_memory: true
    persistent_workers: true
{yaml_otf}

  val:
    name: Auto Val Dataset
    type: pairedimagedataset
    dataroot_gt: [ '{val_hr_dir}' ]
    dataroot_lq: [ '{val_lr_dir}' ]

{network_g_yaml}
{network_d_yaml}

path:
  pretrain_network_g: '{pretrain}'
  strict_load_g: false

train:
  ema_decay: 0.999
  optim_g:
    type: AdamW
    lr: !!float 2e-4
    weight_decay: 0
    betas: [0.9, 0.99]
{optim_d_yaml}

  scheduler:
    type: MultiStepLR
    milestones: {milestones_list}
    gamma: 0.5

  total_iter: {c_total}
  warmup_iter: -1

  losses: {losses_yaml}

val:
  val_enabled: true
  val_freq: 100
  save_img: true
  tile_size: 0
  tile_overlap: 8
  metrics_enabled: false

logger:
  print_freq: 10
  save_checkpoint_freq: 200
  save_checkpoint_format: safetensors
  use_tb_logger: false
"""

config_dir = os.path.join(trainner_dir, "options/train/Easy")
os.makedirs(config_dir, exist_ok=True)

with open(os.path.join(config_dir, f"{project_name}_Phase1.yml"), "w") as f:
    f.write(generate_config(1))
with open(os.path.join(config_dir, f"{project_name}_Phase2.yml"), "w") as f:
    f.write(generate_config(2))

custom_train_code = f"""import os
import sys

if '--auto_resume' not in sys.argv:
    sys.argv.append('--auto_resume')

import warnings
warnings.filterwarnings("ignore")

import logging
logging.disable(logging.WARNING)

import numpy as np
import cv2
cv2.setNumThreads(0)

import torch
from torch.nn import functional as F
from tqdm.auto import tqdm
from traiNNer.check.check_dependencies import check_dependencies

if __name__ == "__main__":
    check_dependencies()

from traiNNer.data import build_dataloader, build_dataset
from traiNNer.data.data_sampler import EnlargedSampler
from traiNNer.data.prefetch_dataloader import CUDAPrefetcher, CPUPrefetcher
from traiNNer.models import build_model
from traiNNer.utils.config import Config
from traiNNer.utils import make_exp_dirs
import glob

GLOBAL_USM_CONFIG = {{
    'enabled': {usm_bool},
    'amount': 0.3,
    'radius': 7,
    'threshold': 10,
    'extra_sharpen_time': 2,
}}

def safe_get(obj, key, default=None):
    if isinstance(obj, dict): return obj.get(key, default)
    return getattr(obj, key, default)

class GPU_APISR_USM(torch.nn.Module):
    def __init__(self, device):
        super().__init__()
        self.device = device
        self.usm_rad = GLOBAL_USM_CONFIG['radius']
        if self.usm_rad % 2 == 0: self.usm_rad += 1
        k1d = cv2.getGaussianKernel(self.usm_rad, 0)
        self.register_buffer('usm_kernel', torch.FloatTensor(np.dot(k1d, k1d.T)).unsqueeze(0).unsqueeze(0))
        sm_k = cv2.getGaussianKernel(3, 0.5)
        self.register_buffer('smooth_kernel', torch.FloatTensor(np.dot(sm_k, sm_k.T)).unsqueeze(0).unsqueeze(0))

    def filter2D_pt(self, img, kernel):
        c, pad = img.size(1), kernel.size(-1) // 2
        return F.conv2d(F.pad(img, (pad, pad, pad, pad), mode='reflect'), kernel.repeat(c, 1, 1, 1), groups=c)

    def apply_apisr(self, img, weight, thresh_val, passes):
        thresh = thresh_val / 255.0
        def usm_pass(x):
            blur = self.filter2D_pt(x, self.usm_kernel)
            res = x - blur
            soft_mask = self.filter2D_pt((torch.abs(res) > thresh).float(), self.usm_kernel)
            return soft_mask * torch.clamp(x + weight * res, 0, 1) + (1.0 - soft_mask) * x
        sharp = usm_pass(img)
        for _ in range(passes): sharp = usm_pass(sharp)
        return torch.clamp(sharp, 0, 1)

    def forward(self, batch):
        if not GLOBAL_USM_CONFIG['enabled'] or 'gt' not in batch:
            if 'lq' in batch:
                batch['lq'] = batch['lq'].to(self.device, non_blocking=True)
            return batch

        gt_ctx = batch['gt'].to(self.device, non_blocking=True)
        gt_ctx = self.filter2D_pt(gt_ctx, self.smooth_kernel)
        gt_ctx = self.apply_apisr(gt_ctx, GLOBAL_USM_CONFIG['amount'], GLOBAL_USM_CONFIG['threshold'], GLOBAL_USM_CONFIG['extra_sharpen_time'])
        batch['gt'] = gt_ctx

        if 'lq' in batch:
            batch['lq'] = batch['lq'].to(self.device, non_blocking=True)

        return batch

def auto_export_onnx(project_name, trainner_dir, current_iter, arch_type="{Model_Architecture}", is_emergency=True):
    import subprocess
    import shutil

    models_dir = os.path.join(trainner_dir, "experiments", project_name, "models")
    search_pattern = f"*net_g*{{current_iter}}.safetensors"
    saved_weights = glob.glob(os.path.join(models_dir, search_pattern))

    if not saved_weights:
        saved_weights = glob.glob(os.path.join(models_dir, "*.safetensors"))

    if saved_weights:
        latest_model = max(saved_weights, key=os.path.getmtime)
        print(f"    ✅ Found native safetensors for export: {{os.path.basename(latest_model)}}")
    else:
        print("    ⚠️ No native safetensors found! Skipping ONNX export.")
        return

    model_path_str = latest_model.replace('\\\\', '/')
    onnx_config = (
        f"name: {{project_name}}\\n"
        f"scale: 2\\n"
        f"num_gpu: auto\\n"
        f"network_g:\\n"
        f"  type: {{arch_type}}\\n"
        f"path:\\n"
        f"  pretrain_network_g: '{{model_path_str}}'\\n"
        f"onnx:\\n"
        f"  dynamo: false\\n"
        f"  opset: 20\\n"
        f"  verify: true\\n"
        f"  optimize: false\\n"
    )

    config_dir = os.path.join(trainner_dir, "options/train/Easy")
    os.makedirs(config_dir, exist_ok=True)
    onnx_config_path = os.path.join(config_dir, "onnx_export_auto.yml")
    with open(onnx_config_path, "w") as f:
        f.write(onnx_config)

    if is_emergency: print(f"\\n🚨 EMERGENCY EXPORT: Converting iter {{current_iter}} checkpoint to ONNX...")
    else: print(f"\\n📦 NORMAL EXPORT: Converting final iter {{current_iter}} checkpoint to ONNX...")

    try:
        result = subprocess.run(
            ["python", os.path.join(trainner_dir, "convert_to_onnx.py"), "-opt", onnx_config_path],
            capture_output=True, text=True, timeout=300
        )
        if result.returncode == 0:
            print("    ✅ Dynamic ONNX export successful!")
            onnx_file = os.path.join(models_dir, f"{{project_name}}.onnx")
            if os.path.exists(onnx_file):
                base_name = project_name.replace("_Phase1", "").replace("_Phase2", "")
                dest_dir = os.path.join(trainner_dir, "experiments", base_name)
                os.makedirs(dest_dir, exist_ok=True)
                dest_path = os.path.join(dest_dir, f"{{base_name}}_{{current_iter}}.onnx")
                shutil.copy(onnx_file, dest_path)
                print(f"    📦 ONNX successfully copied to: {{dest_path}}")
        else:
            print("    ⚠️ ONNX export returned code " + str(result.returncode))
            if result.stderr: print("    Error: " + result.stderr[-500:])
    except Exception as e:
        print("    ❌ ONNX export failed: " + str(e))

def train_pipeline(root_path: str):
    import contextlib
    import io

    opt, args = Config.load_config_from_file(root_path, is_train=True)

    exp_dir = os.path.join(root_path, "experiments", opt.name)
    state_dir = os.path.join(exp_dir, "training_states")
    if os.path.exists(state_dir):
        states = glob.glob(os.path.join(state_dir, "*.state"))
        if states:
            states.sort(key=lambda x: int(os.path.basename(x).split('.')[0]))
            latest_state = states[-1]
            opt.path.resume_state = latest_state

    original_rename = os.rename
    def safe_rename(src, dst):
        if "archived" in dst: return
        original_rename(src, dst)

    os.rename = safe_rename
    with contextlib.redirect_stdout(io.StringIO()):
        try: make_exp_dirs(opt)
        finally: os.rename = original_rename

    if opt.path.resume_state:
        print(f"\\n♻️ Auto-resume active: maintaining current experiment folder.")

    model = build_model(opt)

    current_iter = 0
    if opt.path.resume_state:
        resume_state = torch.load(opt.path.resume_state, map_location="cpu", weights_only=True)
        model.resume_training(resume_state)
        current_iter = resume_state["iter"]

    train_opt = safe_get(opt.datasets, 'train')
    val_opt = safe_get(opt.datasets, 'val')

    train_set = build_dataset(train_opt)
    if len(train_set) == 0: raise ValueError("CRITICAL ERROR: Dataset is completely empty!")

    batch_size = safe_get(train_opt, 'batch_size_per_gpu', 4)
    train_sampler = EnlargedSampler(train_set, 1, 0, max(2000 * batch_size // len(train_set), 1))

    train_loader = build_dataloader(train_set, train_opt, num_gpu=1, dist=False, sampler=train_sampler, seed=1024)

    prefetch_mode = safe_get(train_opt, 'prefetch_mode', 'cpu')
    if prefetch_mode == 'cuda': prefetcher = CUDAPrefetcher(train_loader, opt)
    else: prefetcher = CPUPrefetcher(train_loader)

    val_set = build_dataset(val_opt)
    val_loader = build_dataloader(val_set, val_opt, num_gpu=1, dist=False, sampler=None, seed=1024)

    gpu_processor = GPU_APISR_USM(device=torch.device('cuda')).cuda()

    total_iters = safe_get(opt.train, 'total_iter', 25000)
    print_freq = safe_get(opt.logger, 'print_freq', 10)
    save_freq = safe_get(opt.logger, 'save_checkpoint_freq', 250)
    val_freq = safe_get(opt.val, 'val_freq', 100)
    save_img = safe_get(opt.val, 'save_img', True)

    epoch = 0
    pbar = tqdm(total=total_iters, initial=current_iter, desc=f"🔥 Training {{opt.name}}", unit="iter", dynamic_ncols=True, miniters=1)

    interrupted = False

    try:
        while current_iter < total_iters:
            train_sampler.set_epoch(epoch)
            prefetcher.reset()
            train_data = prefetcher.next()

            while train_data is not None:
                current_iter += 1
                if current_iter > total_iters: break

                with torch.no_grad(): train_data = gpu_processor(train_data)

                model.feed_data(train_data)

                if 'lq' in train_data:
                    model.lq = train_data['lq'].to(model.device)

                model.optimize_parameters(current_iter, 0, True)
                pbar.update(1)

                if current_iter % print_freq == 0:
                    log_dict = model.get_current_log()
                    loss_val = log_dict.get('l_g_total', 0.0) if log_dict else 0.0
                    pbar.set_postfix(Loss=f"{{loss_val:.4f}}", LR=f"{{model.get_current_learning_rate()[0]:.2e}}")

                if current_iter % save_freq == 0 and current_iter > 0:
                    model.save(0, current_iter)
                    pbar.write(f"💾 Checkpoint saved | Iter {{current_iter:,}}")

                if current_iter % val_freq == 0 and current_iter > 0:
                    model.validation(val_loader, current_iter, None, save_img)

                train_data = prefetcher.next()

            epoch += 1

    except (KeyboardInterrupt, RuntimeError) as e:
        interrupted = True
        pbar.close()
        print(f"\\n🛑 INTERRUPTED OR CRASHED at iter {{current_iter}}! Reason: {{type(e).__name__}}")
        print(f"    Saving emergency checkpoint...")

        try:
            model.save(0, current_iter)
            print(f"    ✅ Native State Saved! Next run WILL resume from Iter {{current_iter}}.")
        except Exception as save_err: print(f"    ⚠️ Could not save native state: {{save_err}}")

        auto_export_onnx(opt.name, root_path, current_iter, is_emergency=True)
        print(f"\\n🌟 Emergency export complete! You can safely restart the cell to resume.")
        raise e

    finally:
        if not interrupted:
            pbar.close()
            model.save(-1, -1)
            sys.__stdout__.write(f"\\n✅ Training Phase Complete!\\n")
            sys.__stdout__.flush()
            auto_export_onnx(opt.name, root_path, current_iter, is_emergency=False)

if __name__ == "__main__":
    train_pipeline("{trainner_dir}")
"""

with open(os.path.join(trainner_dir, "OTF_HR_Train.py"), "w") as f:
    f.write(custom_train_code)

script = f"""#!/bin/bash
set -e

PHASE1_DONE={str(phase1_done).lower()}

if [ "$PHASE1_DONE" = "false" ]; then
    echo "=========================================="
    echo "🎬 PHASE 1: L1 Loss | Batch Size: 2 | Iters: {phase1_total}"
    echo "=========================================="
    python "{trainner_dir}/OTF_HR_Train.py" -opt "{config_dir}/{project_name}_Phase1.yml" --auto_resume
fi

echo "=========================================="
echo "🎬 PHASE 2: GAN/Perceptual | Batch Size: 4 | Iters: {phase2_total}"
echo "=========================================="
python "{trainner_dir}/OTF_HR_Train.py" -opt "{config_dir}/{project_name}_Phase2.yml" --auto_resume
"""

with open("run_training.sh", "w") as f:
    f.write(script)

!bash run_training.sh
```

```python
# @title 🖼️/🎥 ONNX Simple Batch Upscaler
# @markdown Point `target_path` to a single file or a folder.
onnx_path = "/content/drive/MyDrive/training ai ups/traiNNer-redux/onnx/Model6kiters.onnx" # @param {type:"string"}
target_path = "/content/Video.mp4" # @param {type:"string"}
auto_download = True # @param {type:"boolean"}

# Internal processing settings
force_resolution = "None"
tile_size = 256
tile_overlap = 16

import os
import sys
import shutil
import subprocess
from glob import glob

try:
    import onnxruntime as ort
except ImportError:
    print("⏳ Installing ONNX Runtime GPU...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "-U", "onnxruntime-gpu"])
    import onnxruntime as ort

import cv2
import numpy as np
from tqdm.auto import tqdm
from google.colab import files

IMAGE_EXTS = {'.jpg', '.jpeg', '.png', '.bmp', '.webp', '.tiff'}
VIDEO_EXTS = {'.mp4', '.avi', '.mkv', '.mov', '.webm', '.flv'}

target_path = os.path.normpath(target_path)
is_dir = os.path.isdir(target_path)

files_to_process = []
if is_dir:
    output_dir = f"{target_path}_upscaled"
    os.makedirs(output_dir, exist_ok=True)
    for ext in IMAGE_EXTS | VIDEO_EXTS:
        files_to_process.extend(glob(os.path.join(target_path, f"*{ext}")))
        files_to_process.extend(glob(os.path.join(target_path, f"*{ext.upper()}")))
else:
    if not os.path.isfile(target_path):
        raise FileNotFoundError(f"❌ Path not found: {target_path}")
    output_dir = os.path.dirname(target_path)
    files_to_process = [target_path]

if not files_to_process:
    raise FileNotFoundError(f"❌ No valid images or videos found in {target_path}")
if not os.path.exists(onnx_path):
    raise FileNotFoundError(f"❌ ONNX file not found: {onnx_path}")

# --- LOAD MODEL ---
print("📦 Loading ONNX Model...")
providers = ['CUDAExecutionProvider', 'CPUExecutionProvider']
session = ort.InferenceSession(onnx_path, providers=providers)

model_inputs = session.get_inputs()
model_outputs = session.get_outputs()
input_name = model_inputs[0].name
is_fp16 = "float16" in model_inputs[0].type

print(f"✅ Model loaded ({session.get_providers()[0]} | {'FP16' if is_fp16 else 'FP32'})")

# Auto-detect scale
try:
    dummy_in = np.zeros((1, 3, 128, 128), dtype=np.float16 if is_fp16 else np.float32)
    dummy_feed = {input_name: dummy_in}
    if len(model_inputs) > 1:
        for inp in model_inputs[1:]:
            d_shape = [1, 1, 128, 128]
            for idx, d in enumerate(inp.shape):
                if isinstance(d, int) and d > 0: d_shape[idx] = d
            dummy_feed[inp.name] = np.zeros(d_shape, dtype=dummy_in.dtype)
    dummy_out = session.run([model_outputs[0].name], dummy_feed)[0]
    scale_factor = max(dummy_out.shape[2] // 128, 1)
except:
    scale_factor = 1
print(f"✅ Auto-detected scale factor: {scale_factor}x\n")

recurrent_states = {}

def apply_force_resolution(frame):
    if force_resolution == "None": return frame
    h, w = frame.shape[:2]
    if force_resolution == "1080p (Stretch)":
        return cv2.resize(frame, (1920, 1080), interpolation=cv2.INTER_CUBIC)
    elif force_resolution == "1080p (Dynamic Aspect Ratio)":
        scale = 1080 / h
        return cv2.resize(frame, (int(w * scale), 1080), interpolation=cv2.INTER_CUBIC)
    return frame

def run_onnx_inference(frame, tile_coord=None):
    global recurrent_states
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB).astype(np.float16 if is_fp16 else np.float32) / 255.0
    in_tensor = np.expand_dims(np.transpose(rgb, (2, 0, 1)), axis=0)
    feed_dict = {input_name: in_tensor}

    if len(model_inputs) > 1:
        for inp in model_inputs[1:]:
            state_key = f"{inp.name}_{tile_coord[0]}_{tile_coord[1]}" if tile_coord else inp.name
            if state_key not in recurrent_states:
                shape = [1 if i==0 or i==1 else (in_tensor.shape[2] if i==2 else in_tensor.shape[3]) if (isinstance(dim, str) or dim is None) else dim for i, dim in enumerate(inp.shape)]
                recurrent_states[state_key] = np.zeros(shape, dtype=np.float16 if is_fp16 else np.float32)
            feed_dict[inp.name] = recurrent_states[state_key]

    result = session.run([out.name for out in model_outputs], feed_dict)

    if len(result) > 1:
        for i in range(1, len(result)):
            if i < len(model_inputs):
                state_key = f"{model_inputs[i].name}_{tile_coord[0]}_{tile_coord[1]}" if tile_coord else model_inputs[i].name
                recurrent_states[state_key] = result[i]

    out_tensor = np.clip(np.squeeze(result[0], axis=0), 0, 1)
    return cv2.cvtColor((np.transpose(out_tensor, (1, 2, 0)) * 255.0).astype(np.uint8), cv2.COLOR_RGB2BGR)

def run_tiled_inference(frame):
    h, w, _ = frame.shape
    t_size = tile_size

    if t_size >= max(h, w):
        return run_onnx_inference(frame)

    stride = t_size - tile_overlap
    pad_h = (t_size - h) if h < t_size else (stride - ((h - t_size) % stride)) % stride
    pad_w = (t_size - w) if w < t_size else (stride - ((w - t_size) % stride)) % stride

    padded = cv2.copyMakeBorder(frame, 0, pad_h, 0, pad_w, cv2.BORDER_REFLECT_101)
    ph, pw = padded.shape[:2]

    accumulator = np.zeros((ph * scale_factor, pw * scale_factor, 3), dtype=np.float32)
    weight_map = np.zeros((ph * scale_factor, pw * scale_factor, 1), dtype=np.float32)

    to_h = to_w = t_size * scale_factor
    tile_weight = np.ones((to_h, to_w, 1), dtype=np.float32)
    t_ov = tile_overlap * scale_factor

    if t_ov > 0:
        for i in range(t_ov):
            v = i / t_ov
            tile_weight[i, :, :] *= v; tile_weight[-1-i, :, :] *= v
            tile_weight[:, i, :] *= v; tile_weight[:, -1-i, :] *= v

    for y in range(0, ph - t_size + 1, stride):
        for x in range(0, pw - t_size + 1, stride):
            out_tile = run_onnx_inference(padded[y:y+t_size, x:x+t_size], tile_coord=(y, x))
            oy, ox = y * scale_factor, x * scale_factor
            accumulator[oy:oy+to_h, ox:ox+to_w] += out_tile.astype(np.float32) * tile_weight
            weight_map[oy:oy+to_h, ox:ox+to_w] += tile_weight

    weight_map[weight_map == 0] = 1.0
    return np.clip(accumulator / weight_map, 0, 255).astype(np.uint8)[0:h*scale_factor, 0:w*scale_factor]

def process_pipeline(frame):
    return apply_force_resolution(run_tiled_inference(frame))

# --- PROCESSING LOOP ---
for idx, file_path in enumerate(files_to_process):
    recurrent_states.clear()
    base_name = os.path.basename(file_path)
    ext = os.path.splitext(base_name)[1].lower()
    out_path = os.path.join(output_dir, f"up_{base_name}")

    print(f"🎬 Processing [{idx+1}/{len(files_to_process)}]: {base_name}")

    if ext in IMAGE_EXTS:
        img = cv2.imread(file_path)
        if img is not None:
            cv2.imwrite(out_path, process_pipeline(img))

    elif ext in VIDEO_EXTS:
        cap = cv2.VideoCapture(file_path)
        fps = cap.get(cv2.CAP_PROP_FPS)
        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

        ret, first_frame = cap.read()
        if not ret: continue

        first_res = process_pipeline(first_frame)
        out_h, out_w = first_res.shape[:2]
        out_h -= out_h % 2
        out_w -= out_w % 2
        first_res = cv2.resize(first_res, (out_w, out_h))

        ffmpeg_cmd = [
            'ffmpeg', '-y', '-hide_banner', '-loglevel', 'error',
            '-f', 'rawvideo', '-vcodec', 'rawvideo',
            '-s', f'{out_w}x{out_h}', '-pix_fmt', 'bgr24', '-r', str(fps),
            '-i', '-', '-i', file_path,
            '-map', '0:v', '-map', '1:a?',
            '-c:v', 'libx264', '-pix_fmt', 'yuv420p', '-crf', '16', '-preset', 'medium',
            '-c:a', 'copy', out_path
        ]

        process = subprocess.Popen(ffmpeg_cmd, stdin=subprocess.PIPE)
        process.stdin.write(first_res.tobytes())

        with tqdm(total=total_frames, desc="Frames", unit="f", leave=False) as pbar:
            pbar.update(1)
            while True:
                ret, frame = cap.read()
                if not ret: break

                res = process_pipeline(frame)
                if res.shape[:2] != (out_h, out_w):
                    res = cv2.resize(res, (out_w, out_h))

                process.stdin.write(res.tobytes())
                pbar.update(1)

        cap.release()
        process.stdin.close()
        process.wait()

print("\n✅ All files processed successfully!")

if auto_download:
    if is_dir:
        print("📦 Zipping outputs for download...")
        zip_path = f"{output_dir}.zip"
        shutil.make_archive(output_dir, 'zip', output_dir)
        files.download(zip_path)
    else:
        files.download(os.path.join(output_dir, f"up_{os.path.basename(files_to_process[0])}"))
```
